````python
Here are the solutions to the assignment questions provided in "Neural Network.pdf". The answers have been formatted to easily copy and paste into a Google Colab notebook, using a mix of Markdown (text) cells for theoretical answers and Python code cells for the implementation tasks.

---

### 📝 Markdown Cell (Text)

**Question 1: What is Deep Learning? Briefly describe how it evolved and how it differs from traditional machine learning.**

**Answer:**

* **Deep Learning:** Deep Learning is a specialized subfield of machine learning that utilizes multi-layered artificial neural networks (hence the term "deep") to extract high-level features from raw data.
* **Evolution:** It began in the 1950s with the invention of the Perceptron (a single-layer neural network). In the 1980s, the introduction of the *backpropagation* algorithm allowed for training multi-layer networks. However, it wasn't until the 2010s that Deep Learning truly exploded, driven by the availability of massive datasets (Big Data) and immense computational power (GPUs).
* **Difference from Traditional ML:** In traditional machine learning, human domain experts must manually perform "feature engineering" to tell the algorithm what to look for (e.g., extracting edge contours from an image before passing it to an SVM). Deep learning automates feature engineering; it feeds raw data (like raw pixels) into the network, and the hierarchical layers automatically learn the optimal features needed to make accurate predictions.

---

### 📝 Markdown Cell (Text)

**Question 2: Explain the basic architecture and functioning of a Perceptron. What are its limitations?**

**Answer:**

* **Architecture & Functioning:** A Perceptron is the simplest form of an artificial neural network. It consists of multiple input nodes, weights, a bias, a summation unit, and an activation function.
1. It takes a set of input values $x_i$.
2. Each input is multiplied by a corresponding weight $w_i$, representing its importance.
3. The weighted inputs are summed together, and a bias $b$ is added to shift the decision boundary. Mathematically: $z = \sum_{i=1}^{n} w_i x_i + b$
4. This weighted sum $z$ is passed through a step activation function to produce a final binary output (e.g., 1 or 0).


* **Limitations:** The fundamental limitation of a single-layer perceptron is that it can only solve **linearly separable** problems. It cannot learn complex, non-linear functions, famously failing to solve the XOR (Exclusive OR) logical problem.

---

### 📝 Markdown Cell (Text)

**Question 3: Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.**

**Answer:**

* **Purpose:** Activation functions introduce *non-linearity* into the neural network. Without an activation function, no matter how many layers a neural network has, it would simply behave like a single-layer linear regression model. Non-linearity allows the network to learn complex patterns and relationships in real-world data.
* **Comparison:**
* **Sigmoid:** Outputs a value between 0 and 1. It is useful for binary classification in the output layer. However, it suffers heavily from the "vanishing gradient" problem (where gradients become too small for the network to learn effectively) and is not zero-centered.
* **Tanh (Hyperbolic Tangent):** Outputs a value between -1 and 1. It is zero-centered, making optimization easier than Sigmoid, but it still suffers from the vanishing gradient problem at its extremes.
* **ReLU (Rectified Linear Unit):** Outputs 0 for negative inputs and the input itself for positive inputs ($f(x) = \max(0, x)$). It is computationally very efficient and solves the vanishing gradient problem for positive values, making it the default choice for hidden layers. However, it can suffer from the "dying ReLU" problem, where negative inputs cause neurons to output zero and stop updating forever.



---

### 📝 Markdown Cell (Text)

**Question 4: What is the difference between Loss function and Cost function in neural networks? Provide examples.**

**Answer:**
While often used interchangeably in casual conversation, technically, they refer to different scopes of error measurement:

* **Loss Function:** Calculates the error for a **single** training example. It measures how far off the model's prediction is from the actual true label for one specific data point.
* *Example:* Binary Cross-Entropy for one specific image classification.


* **Cost Function:** Calculates the average (or sum) of the loss functions over the **entire** training dataset (or a batch). It evaluates how well the model is performing overall.
* *Example:* Mean Squared Error (MSE) averaged across all 10,000 houses in a real estate dataset.



---

### 📝 Markdown Cell (Text)

**Question 5: What is the role of optimizers in neural networks? Compare Gradient Descent, Adam, and RMSprop.**

**Answer:**

* **Role:** Optimizers are algorithms that adjust the attributes of the neural network (specifically, the weights and learning rate) to minimize the Cost Function. They dictate *how* the network learns from its errors.
* **Comparison:**
* **Gradient Descent:** The foundational optimizer. It updates weights by taking a step proportional to the negative of the gradient of the cost function. Standard Gradient Descent uses the whole dataset for one update (computationally slow), while Stochastic Gradient Descent (SGD) uses one sample at a time. It can get stuck in local minima.
* **RMSprop (Root Mean Square Propagation):** An adaptive learning rate method. It divides the learning rate by an exponentially decaying average of squared gradients. This helps the model converge faster by adapting the step size for each weight, making it excellent for Recurrent Neural Networks (RNNs).
* **Adam (Adaptive Moment Estimation):** The most popular and widely used optimizer today. It essentially combines the best properties of RMSprop and Momentum. It computes adaptive learning rates for each parameter by maintaining both a moving average of the gradients (momentum) and the squared gradients (RMSprop).



---

### 💻 Code Cell (Python)

```python
# Question 6: Python program to implement a single-layer perceptron from scratch using NumPy to solve the logical AND gate.[cite: 3]

import numpy as np

# Define the inputs and true outputs for the logical AND gate
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
y = np.array([0, 0, 0, 1])

# Initialize weights, bias, and learning rate
weights = np.random.rand(2)
bias = np.random.rand(1)
learning_rate = 0.1
epochs = 20

# Step activation function
def step_function(z):
    return 1 if z >= 0 else 0

# Training the perceptron
for epoch in range(epochs):
    total_error = 0
    for i in range(len(X)):
        # Forward pass: z = (w1*x1 + w2*x2) + b
        z = np.dot(X[i], weights) + bias
        prediction = step_function(z)

        # Calculate error
        error = y[i] - prediction
        total_error += abs(error)

        # Update weights and bias (Backpropagation)
        weights += learning_rate * error * X[i]
        bias += learning_rate * error

    if total_error == 0:
        print(f"Converged at epoch {epoch+1}")
        break

# Testing the trained perceptron
print("\nTrained Weights:", weights)
print("Trained Bias:", bias)
print("\nTesting AND Gate Predictions:")
for i in range(len(X)):
    z = np.dot(X[i], weights) + bias
    print(f"Input: {X[i]} -> Prediction: {step_function(z)}")

```

---

### 💻 Code Cell (Python)

```python
# Question 7: Implement and visualize Sigmoid, ReLU, and Tanh activation functions using Matplotlib.[cite: 3]

import numpy as np
import matplotlib.pyplot as plt

# Generate input values
x = np.linspace(-10, 10, 200)

# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def tanh(x):
    return np.tanh(x)

# Plotting
plt.figure(figsize=(15, 4))

# Sigmoid Plot
plt.subplot(1, 3, 1)
plt.plot(x, sigmoid(x), color='blue')
plt.title('Sigmoid Activation Function')
plt.grid(True)

# ReLU Plot
plt.subplot(1, 3, 2)
plt.plot(x, relu(x), color='red')
plt.title('ReLU Activation Function')
plt.grid(True)

# Tanh Plot
plt.subplot(1, 3, 3)
plt.plot(x, tanh(x), color='green')
plt.title('Tanh Activation Function')
plt.grid(True)

plt.tight_layout()
plt.show()

```

---

### 💻 Code Cell (Python)

```python
# Question 8: Use Keras to build and train a simple multilayer neural network on the MNIST digits dataset. Print the training accuracy.[cite: 3]

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten

# Load and normalize the MNIST dataset
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# Build the Multilayer Neural Network
model = Sequential([
    Flatten(input_shape=(28, 28)),          # Flatten 2D image to 1D array
    Dense(128, activation='relu'),          # Hidden layer with ReLU
    Dense(64, activation='relu'),           # Second hidden layer
    Dense(10, activation='softmax')         # Output layer (10 classes for digits 0-9)
])

# Compile the model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
print("Training Model...")
model.fit(x_train, y_train, epochs=5, batch_size=32)

# Print final training accuracy
loss, accuracy = model.evaluate(x_train, y_train, verbose=0)
print(f"\nFinal Training Accuracy: {accuracy * 100:.2f}%")

```

---

### 💻 Code Cell (Python)

```python
# Question 9: Visualize the loss and accuracy curves for a neural network model trained on the Fashion MNIST dataset. Interpret the training behavior.[cite: 3]

import tensorflow as tf
import matplotlib.pyplot as plt

# Load Fashion MNIST
fashion_mnist = tf.keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# Build a simple model
model_fashion = tf.keras.models.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

model_fashion.compile(optimizer='adam',
                      loss='sparse_categorical_crossentropy',
                      metrics=['accuracy'])

# Train the model and capture the history object
history = model_fashion.fit(x_train, y_train, epochs=10, validation_data=(x_test, y_test), verbose=0)

# Visualize Loss and Accuracy
plt.figure(figsize=(12, 5))

# Plot Accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend()
plt.grid(True)

# Plot Loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend()
plt.grid(True)

plt.show()

"""
INTERPRETATION OF TRAINING BEHAVIOR:
If the training accuracy steadily increases and training loss decreases, the model is successfully learning.
If the validation accuracy plateau's or validation loss starts to increase while training loss continues to drop,
it indicates that the model is beginning to overfit the training data and losing its ability to generalize to unseen data.
"""

```

---

### 📝 Markdown Cell (Text)

**Question 10: You are working on a project for a bank that wants to automatically detect fraudulent transactions. The dataset is large, imbalanced, and contains structured features like transaction amount, merchant ID, and customer location. The goal is to classify each transaction as fraudulent or legitimate.**

**Explain your real-time data science workflow:**

* **Design a deep learning model:** Because this is structured tabular data with complex, non-linear relationships between features (e.g., location combined with specific amounts), I would design a Multilayer Neural Network (MLP/Feed-forward network). A simple perceptron would fail because fraud patterns are not linearly separable. I would include categorical embeddings for features like `merchant ID` and numerical scaling for `transaction amount`.
* **Activation and Loss Function:**
* **Activation:** I would use **ReLU** for the hidden layers because it is fast, avoids vanishing gradients, and scales well for large datasets. For the output layer, I would use **Sigmoid**, because this is a binary classification problem (Fraud vs. Legitimate) and Sigmoid outputs a probability between 0 and 1.
* **Loss Function:** I would use **Binary Cross-Entropy**. This is the standard mathematical loss function used when predicting probabilities for two mutually exclusive classes.


* **Train and evaluate considering class imbalance:** Fraud datasets typically have 99% legitimate and 1% fraudulent transactions. Using pure "Accuracy" as a metric would be misleading (a model predicting "always legitimate" gets 99% accuracy).
* **Training:** I would use SMOTE (Synthetic Minority Over-sampling Technique) to balance the data, or pass `class_weights` to the Keras `.fit()` function so the model is penalized heavily for missing a fraudulent transaction.
* **Evaluation:** I would evaluate the model using **Precision, Recall (Sensitivity), F1-Score, and the AUC-ROC curve**, prioritizing high Recall to ensure fraudulent transactions are not missed.


* **Optimizer and preventing overfitting:**
* **Optimizer:** I would use the **Adam** optimizer. It adapts the learning rate dynamically, which is highly efficient for large, noisy datasets like financial transactions.
* **Preventing Overfitting:** I would implement **Dropout Layers** (randomly turning off neurons during training to prevent over-reliance on specific features), **Early Stopping** (halting training when validation loss stops improving), and **L2 Regularization**.



---

### 💻 Code Cell (Python)

```python
# Question 10 Python Snippet: Demonstrating Class Weights and Dropout for Fraud Detection[cite: 3]

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# Mock Data Dimensions
number_of_features = 20

# 1. Model Design with Dropout to prevent overfitting
fraud_model = Sequential([
    Dense(64, activation='relu', input_shape=(number_of_features,)),
    Dropout(0.3), # Drops 30% of neurons to prevent overfitting
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid') # Sigmoid for Binary Classification


# 2. Compile using Adam and Binary Cross-Entropy
fraud_model.compile(optimizer='adam',
                    loss='binary_crossentropy',
                    metrics=['accuracy', tf.keras.metrics.Recall(), tf.keras.metrics.AUC()])

# 3. Define Class Weights for Imbalanced Data
# Example: Class 0 (Legit) is heavily prevalent, Class 1 (Fraud) is rare.
# We give Class 1 a much higher weight so the model pays more attention to it.
weights = {0: 1.0, 1: 50.0}

# 4. Define Early Stopping to prevent overfitting
early_stop = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

print("Model compiled successfully. Ready for training with class_weights and callbacks.")
# Example Training Call (assuming X_train, y_train exist):
# fraud_model.fit(X_train, y_train, epochs=20, class_weight=weights, validation_split=0.2, callbacks=[early_stop])

````
